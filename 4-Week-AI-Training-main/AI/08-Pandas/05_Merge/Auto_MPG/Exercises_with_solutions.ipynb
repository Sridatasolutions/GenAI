   ### Step 3. Assign each to a to a variable called cars1 and cars2

```python
cars1 = pd.read_csv(r"C:\Users\srini\GenAI\GenAI\4-Week-AI-Training-main\AI\08-Pandas\05_Merge\Auto_MPG\cars1.csv")
cars2 = pd.read_csv(r"C:\Users\srini\GenAI\GenAI\4-Week-AI-Training-main\AI\08-Pandas\05_Merge\Auto_MPG\cars2.csv")

print(cars1.head())
print(cars2.head())
```

### Step 6. Join cars1 and cars2 into a single DataFrame called cars

```python
cars = pd.concat([cars1, cars2], ignore_index=True)
cars
```
